```python
import pandas as pd
import math
df_tramites = pd.read_excel('./app/data/raw/tramites.xlsx')
df_accesibilidad = pd.read_excel('./app/data/raw/accesibilidad.xlsx')
df_descubrir = pd.read_excel('./app/data/raw/descubrir.xlsx')
```

```python
target_len = 200
df_len = len(df_tramites)
repeats = math.ceil(target_len / df_len)
df_expanded_tramites = pd.concat([df_tramites] * repeats, ignore_index=True).head(target_len)
```

```python
target_len = 200
df_len = len(df_accesibilidad)
repeats = math.ceil(target_len / df_len)
df_expanded_accesibilidad = pd.concat([df_accesibilidad] * repeats, ignore_index=True).head(target_len)
```

```python
target_len = 200
df_len = len(df_descubrir)
repeats = math.ceil(target_len / df_len)
df_expanded_descubrir = pd.concat([df_descubrir] * repeats, ignore_index=True).head(target_len)
```

```python
print(len(df_expanded_tramites))
print(len(df_expanded_accesibilidad))
print(len(df_expanded_descubrir))
```

```python
df_total = pd.concat([df_expanded_tramites, df_expanded_accesibilidad, df_expanded_descubrir], ignore_index=True)
```

```python
df_total.iloc[:, [1]].to_csv("output.csv", index=False)
```

```python
responses = []
import json
response_file_path = './app/data/processed/load/test_6000.json'
with open(response_file_path, 'r', encoding='UTF-8') as f:
  responses = json.load(f)
```

```python
print(len(responses))
```

```python
to_csv = []
for i in responses:

  response_time = i.get('response_time', 0)

  partial = (i.get('partial_answers') or {})
  router = partial.get('router')
  if router is not None:
    route = router.get('route','undefined')

  grounding = partial.get('grounding', {})
  if grounding is not None:
    grounded = grounding.get('is_grounded', False)
    
  item = [response_time, route, grounded]
  to_csv.append(item)
```

```python
print(len(to_csv))
```

```python
import pandas as pd
df = pd.DataFrame(to_csv, columns=["response_time", "route", "reach_grounded"])
df.to_csv("data_6000.csv", index=False)
```
